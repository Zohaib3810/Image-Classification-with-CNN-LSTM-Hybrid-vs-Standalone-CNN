```python
from tensorflow.keras.datasets import cifar10
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Load CIFAR-10 dataset
(x_train_c, y_train_c), (x_test_c, y_test_c) = cifar10.load_data()
x_train_c, x_test_c = x_train_c / 255.0, x_test_c / 255.0

# 2. Design CNN-LSTM Hybrid Model with Normalization
hybrid_model_v2 = models.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    # Output of pool2 is (6, 6, 64). Reshape to (36, 64) for LSTM
    layers.Reshape((6 * 6, 64)),
    layers.LSTM(64),
    layers.BatchNormalization(),
    layers.Dense(10, activation='softmax')
])

# 3. Standalone CNN with Normalization
cnn_model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.BatchNormalization(),
    layers.Dense(10, activation='softmax')
])

# Compile and Train Hybrid
hybrid_model_v2.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("\nTraining Task 2: CNN-LSTM Hybrid with Normalization")
hybrid_model_v2.fit(x_train_c, y_train_c, epochs=5, batch_size=64, validation_data=(x_test_c, y_test_c))

# Compile and Train Standalone CNN
cnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("\nTraining Task 2: Standalone CNN with Normalization")
cnn_model.fit(x_train_c, y_train_c, epochs=5, batch_size=64, validation_data=(x_test_c, y_test_c))
```
